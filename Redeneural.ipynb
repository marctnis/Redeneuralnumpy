```python
import os
import numpy as np
from PIL import Image
import kagglehub
from tqdm import tqdm
```

```python
path = kagglehub.dataset_download("olafkrastovski/handwritten-digits-0-9")

lista_imagens = []
lista_rotulos = []

pastas = [p for p in os.listdir(path) if os.path.isdir(os.path.join(path, p))]

for pasta_digito in tqdm(pastas, desc="Processando pastas (0-9)"):
    caminho_pasta = os.path.join(path, pasta_digito)

    for nome_arquivo in os.listdir(caminho_pasta):
        caminho_imagem = os.path.join(caminho_pasta, nome_arquivo)

        try:
            img = Image.open(caminho_imagem)
            lista_imagens.append(np.array(img))
            lista_rotulos.append(int(pasta_digito))
        except Exception:
            continue

X = np.array(lista_imagens, dtype=object)
y = np.array(lista_rotulos)

print(f"\nCarregamento concluído!")
```
