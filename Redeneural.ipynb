```python
import urllib.request
import numpy as np

# URL direta para o arquivo de pixels (mfeat-pix) no repositório UCI
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/mfeat/mfeat-pix"

print("Baixando e processando o dataset...")

# Faz a requisição e lê os dados diretamente para o NumPy
with urllib.request.urlopen(url) as response:
    # O arquivo tem 2000 linhas e 240 colunas (características dos pixels)
    X = np.loadtxt(response)

# Cria os rótulos (y) já que o dataset é ordenado (200 imagens por dígito de 0 a 9)
y = np.repeat(np.arange(10), 200)

print(f"Dataset carregado com sucesso!")
print(f"Dimensão das features (X): {X.shape} -> (2000 amostras, 240 pixels)")
print(f"Dimensão dos rótulos (y): {y.shape}")
```

```python
import numpy as np

indices = np.arange(X.shape[0])
np.random.shuffle(indices)

X_embaralhado = X[indices]
y_embaralhado = y[indices]

X_split = np.array_split(X_embaralhado, 3)
y_split = np.array_split(y_embaralhado, 3)

X_treino = np.concatenate((X_split[0], X_split[1]))
y_treino = np.concatenate((y_split[0], y_split[1]))

X_teste = X_split[2]
y_teste = y_split[2]
```

```python
import numpy as np

def codificar_one_hot(rotulos, total_classes=10):
    matriz_one_hot = np.zeros((rotulos.size, total_classes))

    matriz_one_hot[np.arange(rotulos.size), rotulos] = 1

    return matriz_one_hot

y_treino_oh = codificar_one_hot(y_treino)
y_teste_oh = codificar_one_hot(y_teste)

print("Formato do y_treino original:", y_treino.shape)
print("Formato do y_treino em One-Hot:", y_treino_oh.shape)

print("\nExemplo da primeira amostra:")
print(f"Gabarito original: {y_treino[0]}")
print(f"Vetor One-Hot:     {y_treino_oh[0]}")
```
